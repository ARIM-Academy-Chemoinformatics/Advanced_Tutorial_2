### フラグメント化（部分構造化）

RDKitの`BRICS.BRICSDecompose()`メソッドは、分子をBRICS（Breaking of Retrosynthetically Interesting Chemical Substructures、本編冒頭を参照）法に基づいてフラグメント化するための機能を提供します。

《注意》BRICSは「Bristol-Myers Squibb」とは無関係です。逆合成的に切断可能な結合（Retrosynthetically Interesting）を切る、という命名です（Degen et al., 2008）。
`BRICSDecompose()`によって、元の分子が複数のフラグメントに分割され、それぞれのフラグメントが独立した化学構造となります。なお、デフォルトでは切断位置にダミーアトムを挿入して表示されます。

化合物IDゼロ番目をフラグメント化させると、6つに分解されることがわかります。

《注意：ダミー原子の数字（R型ラベル）について》上の切断結果に現れる`[16*]`や`[13*]`のような数字付きのダミー原子（`*`）は、BRICSが定義する**16種類の化学的環境（結合タイプ）のカテゴリ番号（R型ラベル、L1〜L16）**を表しています。各カテゴリはSMARTSパターンで定義された特定の化学的文脈（例：`L16`は芳香環の炭素、`L1`はアミド/エステルのカルボニル炭素）に対応します。BRICSは、全ての番号どうしを無差別に結合できるわけではなく、**あらかじめ定義された特定の番号の組み合わせ（例：`L1`と`L3`、`L3`と`L16`など）のみを結合可能**とするルールを持っています。このルールにより、化学的に不自然な結合ではなく、実際の逆合成（レトロシンセシス）で妥当とされる結合様式に基づいた仮想分子が生成されます（詳細はDegen et al., 2008を参照）。

### フラグメントライブラリ作成
上記は1分子でのフラグメントの方法でした。次はdelaney-processed.csvにある1128分子のフラグメント化を行います。1128化合物の全フラグメントを重複のないリスト形式としてまとめライブラリ化します。

まず、三項演算子を使い、df['mol']を、逐次、`BRICSDecompose()`関数へmolという変数で受け渡し`list()`関数でリスト配置をとらせます。

````python
fragments = [list(BRICS.BRICSDecompose(mol)) for mol in df['mol']]
````

````python
len(fragments)
````

このようにフラグメントは1128分子についてネスト化されたリスト構造となっています。ネストを解消させてフラグメントのリスト構造とする方法として、次のような`unwrap`関数を準備します。

````python
def unwrap(list_data):
    list_output = []
    for li in list_data: 
        list_output += li
    return list_output
````

unwrap関数にかけると、フラグメントの総数は2315のパーツであることがわかります。

````python
fr_all = unwrap(fragments)
````

````python
len(fr_all)
````

フラグメントでは重複するものがあります。ここでは`set()`関数を使って、一意のフラグメントの集合をつくります。下記のように実行するとフラグメント数として1028となりました。すなわち約半分が重複していたことになります。

````python
# 【修正】set()の反復順序はPYTHONHASHSEEDに依存し、実行のたびに変わりうる（実行環境間で再現しない）。
# 後続のBRICSBuild()に渡すcomponentの並び順を固定するため、sorted()で決定的な順序にしてから集合化する。
frag_library = sorted(set(fr_all))
````

````python
print('BRICSが取り出した全フラグメント', len(frag_library))
````

### 部分構造の指定
フラグメント化された構造の一部を確認しましょう。frag_libraryのSMILESをMolオブジェクトに変換したものをlist_fragmentsに格納します。

````python
# 【修正】frag_libraryの先頭2つを機械的に選ぶと、結合可能な部位（ダミー原子）を持たない
# フラグメントや、特殊な原子のみからなる小さな断片（例：臭素だけのフラグメント）が選ばれ、
# 後続のBRICSBuild(seeds=...)で有効な組み合わせがほとんど得られない（生成数が0近くになる）場合がある。
# ここでは「ダミー原子（結合可能部位）を持ち、C・N・Oのみからなる中程度の大きさ
# （重原子数6〜14）」のフラグメントに絞り込んでから、先頭2つをシード構造として選ぶ。
list_fragments = [Chem.MolFromSmiles(smi) for smi in frag_library]

candidate_seeds = []
for m in list_fragments:
    if m is None:
        continue
    n_dummy = sum(1 for a in m.GetAtoms() if a.GetAtomicNum() == 0)
    symbols = set(a.GetSymbol() for a in m.GetAtoms() if a.GetAtomicNum() != 0)
    if n_dummy >= 1 and symbols <= {'C', 'N', 'O'} and 6 <= m.GetNumHeavyAtoms() <= 14:
        candidate_seeds.append(m)

seed_structures = candidate_seeds[:2]
print(len(seed_structures))
````

絞り込んだフラグメントの最初の2つの構造を描画したものを表示させます。上の【修正】によりfrag_libraryの並び順は固定されているため、実行ごとに同じ構造が選ばれます。

````python
# 図3：フラグメントライブラリの先頭2構造（frag_libraryの並び順に依存）
Draw.MolsToGridImage(seed_structures)
````

### 構造生成によるバーチャルライブラリーの構築

生成されたフラグメントを利用して元の化合物以外の新しい化合物を生成し、「仮想ライブラリ」を作成します。

**構造生成（Structure Generation）**:  
構造生成は、特定の条件や性質を満たす新しい化合物をコンピュータ上で生成するプロセスを指します。化学的に妥当な構造を効率的に生成することは、薬物設計や材料科学などの分野で重要です。フラグメントを組み合わせて新しい化合物の構造生成が行えますが、`BRICS.BRICSBuild()`関数を用いると、部分構造化されたフラグメントをコンピューター上で機械的に組み合わせて新しい構造を生成することが簡単にできます。

**仮想ライブラリ（Virtual Library）**:  
仮想ライブラリとは、化学的に実際に合成されていないが、理論的には存在し得る全ての化合物の集合を指します。仮想ライブラリを用いることで、実験室で合成するのは困難な大規模な化合物集合を網羅的に探索し、有望な化合物を同定することが可能です。また、仮想ライブラリを使うことで、既知の活性を持つ化合物の拡張や、新しい化合物の探索が進められます。

### BRICSBuild()のデフォルト設定での構造生成
frag_libraryのSMILESをMolオブジェクトに変換し、component変数として格納します。続いて`BRICSBuild()`関数で機械的に構造生成を行いbuild変数に格納します。この１行の処理で仮想ライブラリの荒削りの仮想ライブラリが生成されます。 

````python
component = [Chem.MolFromSmiles(smi) for smi in frag_library]
build = BRICS.BRICSBuild(component) 
````

````python
build
````

【解説】`BRICS.BRICSBuild()` メソッド

`BRICS.BRICSBuild()` メソッドは、フラグメントされたMolオブジェクトのリストから、新しい化合物を生成する方法を提供します。
使用法は以下のようになります：

```python
BRICS.BRICSBuild(mol seeds = seed_structures, maxDepth = 3)
```

引数:
- `mol`: RDKitのMolオブジェクトで、分割対象となる化合物を表します。
- `seeds` (オプション): 構造生成で必ず含むフラグメントを指定します。
- `maxDepth` (オプション): 結合回数を指定します。

戻り値:
`BRICS.BRICSBuild()` は、新しい構造生成の化合物からなるMolオブジェクトのジェネレータを返します。


分割された部分構造を統合することで、元の化合物とは異なる新しい化合物が得られます。

上記の`build`に収まっている構造生成された化合物は、**ジェネレータ（generator）型**で保持されています。`BRICSBuild()`は、考えられる組み合わせを一度に全てメモリ上に作るのではなく、`next()`が呼ばれるたびに新しい分子を1つずつその場で生成します（大量の組み合わせを扱う場合にメモリを節約できる仕組みです）。そのため`build`自体を表示しても`<generator object ...>`としか表示されず、実際に分子を取り出すには次のように`next()`関数を使って逐次的に呼び出してゆきます。

このとき、`UpdatePropertyCache(strict=True)`を呼び出しています。フラグメントを機械的に結合させると、価数（結合数）の辻褄が合わない化学的に不自然な構造（例：5本の結合を持つ炭素原子など）が生成される場合があります。`UpdatePropertyCache(strict=True)`は、RDKitの内部で管理している原子の価数・水素数などのキャッシュ情報を再計算し、こうした化学的に不適切な構造が混入していないかをチェックするために実行しています。チェックを通ったものを`append()`メソッドで`generated_mols`に格納しています。このように集められた`generated_mols`が、新しく構造生成された分子集合＝「仮想ライブラリ」となります。


````python
### 分子1000個を作成
generated_mols = []

for i in range(1000):
    m = next(build)
    m.UpdatePropertyCache(strict=True)
    generated_mols.append(m)
````

得られた1000化合物の仮想化合物のうち、90化合物を表示してみましょう。

````python
# 図4：BRICSBuild()により新規生成した仮想化合物（ランダムに90個抽出）
np.random.seed(1234)
np.random.shuffle(generated_mols)

Draw.MolsToGridImage(generated_mols[:90], molsPerRow=3, subImgSize=(300,150))
````

````python
### 分子1000個を作成
generated_mols = []
for i in range(1000):
    m = next(build2)
    m.UpdatePropertyCache(strict=True)
    generated_mols.append(m)
````

## 3.構造生成とファイル保存　(底本P.125)

底本では構造生成数を10万としていますが、本編では実行時間短縮のため3千（3,000）に変更しています。構造生成を逐次処理とmultiprocessingによる並列処理とで行い、その処理時間について`time`関数を使って比較を行います。

《注意》multiprocessingの挙動はOSに依存します。Linux（Google Colab、および多くのローカルLinux環境を含む）では既定のfork方式が使われるため、本編のようにセル内で定義した関数でも並列処理は正常に動作します（本編はLinux環境で実際に動作確認済みです）。一方、Windows・macOS（Python 3.8以降）では既定がspawn方式となり、対象関数を独立したモジュールとして`.py`ファイルに定義しておく必要がある場合があります。お使いの環境で動作しない場合は、この違いを疑ってみてください。

````python
from random import seed
from time import time
from multiprocessing import Pool
````

````python
def stopwatch(t):
    """経過時間を測る関数"""
    return (time() - t)/60
````

### 逐次処理

````python
import os
os.makedirs('results', exist_ok=True)  # ローカル実行では results/ フォルダが存在しない場合があるため作成する

NUM_ITER=3000

start = time()
seed(20200315)

builder = BRICS.BRICSBuild(component) 

with open('./results/mol_single.smi', 'w') as f:
    for i in range(NUM_ITER):
        m = next(builder)
        m.UpdatePropertyCache(strict=True)
        
        smi = Chem.MolToSmiles(m)
        f.write(smi+'\n')
        
print('Elapsed time', stopwatch(start), '[mins]')
````

### 並列処理
留意：OSによって挙動が異なります（上の《注意》を参照）。うまく動作しない場合はセルの再実行やカーネルの再起動を試してください。

````python
def sample_molecule(mol):
    """
    multiprocessing.Pool.imapに渡す関数。
    Molオブジェクトの更新をしたあとでSMILESに変換。
    """
    mol.UpdatePropertyCache(strict=True)
    return Chem.MolToSmiles(mol)
````

````python
NUM_ITER=3000

start2 = time()
seed(20200315)

c = 0

builder = BRICS.BRICSBuild(component) 

with Pool(4) as p:
    f = open('./results/mol_quad.smi', 'w')
    
    for smi in p.imap(func=sample_molecule, iterable=builder, chunksize=100):
        f.write(smi+'\n')
        c+=1
        if c == NUM_ITER:
            print(c)
            break
    f.close()
    
print('Elapsed time', stopwatch(start2), '[mins]')
````

## ①逐次処理と並列処理の比較

同じ1万個の仮想化合物生成タスクを、逐次処理（シングルプロセス）とmultiprocessing（4プロセス）で実行し、所要時間を比較します。


````python
# ①逐次処理と並列処理の所要時間を比較する
elapsed_single = stopwatch(start)
elapsed_quad = stopwatch(start2)

comparison = pd.DataFrame({
    '処理方式': ['逐次処理（シングルプロセス）', '並列処理（4プロセス, Pool(4)）'],
    '生成分子数': [NUM_ITER, NUM_ITER],
    '所要時間[分]': [elapsed_single, elapsed_quad],
})
comparison
````

- `set()`はPythonの実行のたびに反復順序が変わりうるため、`random.seed()`を設定しても生成される化合物の並びまでは再現できないことを確認しました（本編の【修正】箇所を参照）。乱数シードだけでなく、集合や辞書の順序に依存する処理がないかにも注意が必要です。
- 逐次処理と並列処理（4プロセス）の所要時間を①の比較表で比較しました。プロセス生成のオーバーヘッドがあるため、化合物1個あたりの処理が軽い場合は、並列化してもコア数倍の高速化が得られるとは限りません。

## ③本ノートブックで扱っていないこと（今後の課題）

- 生成された仮想化合物が化学的に妥当か（合成可能性、Lipinskiのルールオブファイブなど）のフィルタリングは行っていません。
- 生成した仮想化合物に対する物性予測（3_Solubility.ipynbのモデルの適用）は5_Screening.ipynbで扱います。
- `BRICSBuild()`の`maxDepth`パラメータを変えたときの生成される化合物の複雑さの違いは詳しく検討していません。
